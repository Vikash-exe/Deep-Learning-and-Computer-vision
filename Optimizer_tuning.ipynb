```python
import keras_tuner as kt
```

```python
def model_build(hp):
    model = Sequential([
        Dense(32, activation = "relu", input_shape = (x_train.shape[1],)),
        Dense(1, activation = "sigmoid")
    ])

    optimizers = hp.Choice('optimizers', values = ["Adam", "SGD", "RMSprop", "Adagrad", "Adadelta", "Adamax", "Nadam"])
    model.compile(optimizer = optimizers, loss = "binary_crossentropy", metrics = ["accuracy"])
    return model
```

```python
tuner = kt.RandomSearch(
    model_build,
    objective = "val_accuracy",
    max_trials = 5,
    overwrite = True
)
```

```python
tuner.search(x_train, y_train, epochs = 10, batch_size = 32, validation_data=(x_test, y_test), verbose=2)
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
model.fit(x_train, y_train, epochs = 100,initial_epoch = 11, batch_size = 32, validation_data=(x_test, y_test))
```

```python
def model_builder(hp):
    model = Sequential([
        Dense(units = hp.Int( 'units' ,min_value=32, max_value=512, step=32), activation = "relu", input_shape = (x_train.shape[1],)),
        Dense(1, activation = "sigmoid")
    ])

    model.compile(optimizer = "rmsprop", loss = "binary_crossentropy", metrics = ["accuracy"])
    return model
```

**For number of hidden layers in model**

```python
def model_builder(hp):
    model = Sequential()
    model.add(Dense(32, activation = "relu", input_shape = (x_train.shape[1],)))
    for i in range(hp.Int('num_layers', 1, 5)):
        model.add(Dense(32, activation = "relu"))
    model.add(Dense(1, activation = "sigmoid"))

    model.compile(optimizer = "rmsprop", loss = "binary_crossentropy", metrics = ["accuracy"])
    return model
```

```python
tuner = kt.RandomSearch(
    model_builder,
    objective = "val_accuracy",
    max_trials = 5,
    overwrite = True,
    directory = "F:/Python/Deep-Learning-and-Computer-vision/Optimizer_tuning",
    project_name = "num_layers_tuning"
)
```

```python
tuner.search(x_train, y_train, epochs = 10, batch_size = 32, validation_data=(x_test, y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

**ALL TOGETHER**

```python
def build_model_all(hp):
    model = Sequential()

    counter = 0

    for i in range(hp.Int('num_layer', min_value=1, max_value=10)):

        if counter == 0:
            model.add(Dense(units = hp.Int('units_' + str(i), min_value=8, max_value=128, step=8),
                             activation = hp.Choice('activation_'+str(i), values= ['relu', 'tanh', 'sigmoid']), 
                             input_shape = (x_train.shape[1],)))
            model.add(Dropout(rate = hp.Float('dropout_' + str(i), min_value=0.0, max_value=0.5, step=0.1)))
        else:
            model.add(Dense(units = hp.Int('units_' + str(i), min_value=8, max_value=128, step=8),
                             activation = hp.Choice('activation_'+str(i), values= ['relu', 'tanh', 'sigmoid'])))
            model.add(Dropout(rate = hp.Float('dropout_' + str(i), min_value=0.0, max_value=0.5, step=0.1)))
    counter += 1
    model.add(Dense(1, activation = "sigmoid"))
    model.compile(optimizer = hp.Choice('optimizer', values = ['adam', 'sgd', 'rmsprop']),
                   loss = "binary_crossentropy",
                    metrics = ["accuracy"])
    return model
```

```python
tuner =kt.RandomSearch(
    build_model_all,
    objective='val_accuracy',
    max_trials=5,
    directory = "F:/Python/Deep-Learning-and-Computer-vision/Optimizer_tuning",
    project_name = "All_param_tuning"
)
```

```python
tuner.search(x_train, y_train, epochs = 5, validation_data=(x_test, y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train, y_train, epochs = 100, initial_epoch = 5, batch_size = 32, validation_data=(x_test, y_test))
```
