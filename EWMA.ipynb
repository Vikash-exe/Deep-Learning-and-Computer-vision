```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('F:\Python\Deep-Learning-and-Computer-vision\Dataset\DailyDelhiClimateTrain.csv')
```

```python
df.head()
```

```python
df = df[['date', 'meantemp']]
```

```python
df = df.head(100)
```

```python
plt.scatter(df['date'], df['meantemp'])
plt.show()
```

```python
x1 = df['meantemp'].ewm(alpha=0.1).mean()
```

```python
df = df.assign(x1=x1)
```

```python
df.head()
```

```python
plt.plot(df['date'], df['meantemp'], label='Original')
plt.plot(df['date'], df['x1'], label='EWMA', color='red')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Mean Temperature', fontsize=12)
plt.xticks(rotation=45, fontsize=8)
plt.yticks(fontsize=8)
plt.legend()
plt.show()
```
