```python
import tensorflow as tf
```

```python
tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9, nesterov=True, name='NAG') #for Nesterov Accelerated Gradient (NAG)
```

```python
tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.0, nesterov=False, name='SGD') #for Stochastic Gradient Descent (SGD)
```

```python
tf.keras.optimizers.Adagrad(learning_rate=0.01, name='AdaGrad') #for Adaptive Gradient (AdaGrad)
```

```python
tf.keras.backend.clear_session() #for clearing the previous models and layers from memory to avoid clutter and potential conflicts when creating new models.
```
